# 데이터 셋 구조 

- text: 파싱된 텍스트 전체
- labels: 정답 json의 경로
- original_path: 원본 pdf의 경로(실제 사용되지 않음)


추가로 output_sheme.py에 pydantic 만들고, import하고, 코드 수정하기 

```python
import pandas as pd
```

```python
def add_data(prev_df, md_path, json_path, file_path):
    if prev_df is None:
        prev_df = pd.DataFrame(columns=["text", "labels", "original_path"])
    md = open(md_path, 'r').read()
    new_row = pd.DataFrame({
        "text": [md],
        "labels": [json_path],
        "original_path": [file_path]
    })
    return pd.concat([prev_df, new_row], ignore_index=True)
```

```python
new_data = add_data(None,
                    "/home/bch/Project/main_project/scout_partners/Resume-analysis/result/250619_124206/modified_result.md",
                    "/home/bch/Project/main_project/scout_partners/llm-ner-benchmarks/data/json/사람인s3경력.json",
                    "/home/bch/Project/main_project/scout_partners/Resume-analysis/result/250619_124206/잡코리아-s3.pdf"
                    )
new_data.head()
```
